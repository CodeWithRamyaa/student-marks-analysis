```python
import pandas as pd

def load_data(filename):
    return pd.read_csv(filename)

def analyze_marks(df):
    subjects = ["Maths", "Science", "English"]

    df["Average"] = df[subjects].mean(axis=1)

    print("STUDENT PERFORMANCE")
    print(df.to_string(index=False))

    print("\nClass average:", round(df["Average"].mean(), 2))
    print("Highest average:", round(df["Average"].max(), 2))
    print("Lowest average:", round(df["Average"].min(), 2))

    return df

def save_data(df, filename):
    df.to_csv(filename, index=False)
    print("\nData saved to", filename)

# Run the project
df = load_data("student_marks.csv")
df = analyze_marks(df)
save_data(df, "student_analysis_report.csv")
```

```python
import pandas as pd

def calculate_average(marks):
    return sum(marks) / len(marks)

def get_grade(average):
    if average >= 80:
        return "A"
    elif average >= 60:
        return "B"
    elif average >= 40:
        return "C"
    else:
        return "F"

# Test 1: Average calculation
assert calculate_average([80, 70, 90]) == 80

# Test 2: Grade classification
assert get_grade(85) == "A"
assert get_grade(70) == "B"
assert get_grade(50) == "C"
assert get_grade(30) == "F"

# Test 3: Student data
students = {
    "Anu": [85, 88, 80],
    "Ravi": [72, 75, 68]
}

assert calculate_average(students["Anu"]) == 253 / 3
assert get_grade(72) == "B"

print("All tests passed successfully!")
```

```python
import unittest

def calculate_average(marks):
    return sum(marks) / len(marks)

def get_grade(average):
    if average >= 80:
        return "A"
    elif average >= 60:
        return "B"
    elif average >= 40:
        return "C"
    else:
        return "F"

class TestStudentMarks(unittest.TestCase):

    def test_average(self):
        self.assertEqual(calculate_average([80, 70, 90]), 80)

    def test_grade_A(self):
        self.assertEqual(get_grade(85), "A")

    def test_grade_B(self):
        self.assertEqual(get_grade(70), "B")

    def test_grade_C(self):
        self.assertEqual(get_grade(50), "C")

    def test_grade_F(self):
        self.assertEqual(get_grade(30), "F")

suite = unittest.TestLoader().loadTestsFromTestCase(
    TestStudentMarks
)

result = unittest.TextTestRunner(verbosity=2).run(suite)

print("\nAll tests passed:", result.wasSuccessful())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Student dataset
data = {
    "Student": ["Anu", "Ravi", "Priya", "Kiran", "Sneha"],
    "Maths": [85, 72, 91, 34, 78],
    "Science": [88, 75, 95, 45, 82],
    "English": [80, 68, 89, 39, 76]
}

df = pd.DataFrame(data)

# Calculate each student's average
subjects = ["Maths", "Science", "English"]
df["Average"] = df[subjects].mean(axis=1)

# Display the report
print("STUDENT PERFORMANCE REPORT")
print(df.to_string(index=False))

# Create a bar chart
plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Average"])
plt.title("Average Marks by Student")
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.tight_layout()

# Save the chart
plt.savefig("student_performance.png")
plt.show()

print("Chart saved successfully!")
```
